### Importa bibliotecas

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

### Carrega o dataset

```python
bullying = pd.read_csv('dados/bullying.csv', sep=';')
```

### Configura os gráficos

```python
plt.rcParams['figure.figsize']=(17,8.5)
sns.set_style("whitegrid")
```

### Locais

```python
locais = bullying['local'].value_counts()
```

```python
sns.barplot(x=locais.index, y=locais.values, alpha=0.8);
plt.title("Entrevistado por Locais");
```
